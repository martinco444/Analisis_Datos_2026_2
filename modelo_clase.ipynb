importar librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
from pathlib import Path
from sklearn.preprocessing import OneHotEncoder
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

obtener el dataset

```python
# Download latest version
path = kagglehub.dataset_download("gandpablo/cars-prices-for-machine-learning")

print("Path to dataset files:", r"C:\dev\Analisis_Datos_2026_2")
```

mostrar el dataset

```python
archivos = list(Path(path).rglob("*"))
print("Archivos encontrados:")
for archivo in archivos:
    if archivo.is_file():
        print(archivo.name)

archivo_csv = next(archivo for archivo in archivos if archivo.suffix.lower() == ".csv")
df = pd.read_csv(archivo_csv)

display(df.head())
```

dividir el conjunto de datos X - y

```python
# Eliminar filas y columnas completamente vacías
df = df.dropna(axis=0, how="all").dropna(axis=1, how="all")

# Eliminar columna de índice original
df = df.drop(columns=["Unnamed: 0"], errors="ignore")

# Eliminar filas con valores nulos
df = df.dropna().copy()

# Separar primero la variable objetivo
y = df["Transmission"].copy()
X = df.drop(columns=["Transmission"]).copy()

# Convertir variables booleanas de X a 0 y 1
columnas_booleanas = X.select_dtypes(include="bool").columns
X[columnas_booleanas] = X[columnas_booleanas].astype(int)

# Convertir solamente las variables categóricas de X
columnas_categoricas = X.select_dtypes(
    include=["object", "category"]
).columns

X = pd.get_dummies(
    X,
    columns=columnas_categoricas,
    dtype=int
)

display(X.head())
print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
print("Clases:", y.unique())
```

particionar

```python
#particionar X y
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
modelo = svm.SVC(kernel='linear')
modelo = modelo.fit(X_train, y_train)
print(modelo)
```

```python
y_pred = modelo.predict(X_test)
print(classification_report(y_test, y_pred))
```
